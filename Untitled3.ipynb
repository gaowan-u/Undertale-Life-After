```python

# 第一步：安装我们需要的库 (在Colab环境中这会非常快且不会报错)
!pip install opencv-python-headless numpy Pillow

# 第二步：粘贴我们之前最终版的OpenCV代码
import cv2
import numpy as np
from PIL import Image
import os

def split_elements_with_opencv(image_path, element_name="element"):
    print(f"--- 正在使用 OpenCV 处理: {image_path} ---")
    img_pil = Image.open(image_path).convert("RGBA")
    img_np = np.array(img_pil)
    alpha_channel = img_np[:, :, 3]
    contours, _ = cv2.findContours(alpha_channel, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    print(f"在 {image_path} 中检测到 {len(contours)} 个独立的元素。")
    if not contours:
        print("未找到任何可处理的元素。\n")
        return []
    elements_found = []
    for i, contour in enumerate(contours):
        x, y, w, h = cv2.boundingRect(contour)
        bbox = (x, y, x + w, y + h)
        position = (x, y)
        cropped_img = img_pil.crop(bbox)
        output_path = f"cropped_{element_name}_{i+1}.png"
        cropped_img.save(output_path)
        print(f" - 元素 {i+1} ({element_name}) 已裁剪并保存至: {output_path}")
        print(f" - 它在原图 '{image_path}' 中的相对位置是: {position}")
        elements_found.append({
            "image": cropped_img,
            "position": position,
            "name": f"{element_name}_{i+1}"
        })
    print("-" * 25 + "\n")
    return elements_found

# --- 主逻辑 ---
try:
    background = Image.open("出生点.png").convert("RGBA")
    final_image = background.copy()
    layers = {
        '摇杆底座.png': 'joystick_base',
        '摇杆实体.png': 'joystick_top',
        '按钮.png': 'button'
    }
    for layer_path, name in layers.items():
        if not os.path.exists(layer_path):
            print(f"警告：文件 '{layer_path}' 不存在，已跳过。\n")
            continue
        individual_elements = split_elements_with_opencv(layer_path, name)
        for element in individual_elements:
            element_img = element["image"]
            final_position = element["position"]
            print(f"粘贴元素 '{element['name']}' 从 '{layer_path}' 到最终画布位置: {final_position}")
            final_image.paste(element_img, final_position, element_img)
    final_image.save("final_composite_image_opencv.png")
    print("\n所有元素已成功分离并精确合成！最终图片已保存为 'final_composite_image_opencv.png'")
except FileNotFoundError:
    print("\n错误：找不到背景文件 '出生点.png'。请确保所有图片文件都已上传。")
except Exception as e:
    print(f"\n程序运行出错: {e}")
```
